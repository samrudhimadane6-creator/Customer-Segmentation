```python
import numpy as np

grid = np.zeros((4,4))

goal = (3,3)

value = np.zeros((4,4))

gamma = 0.9

for iteration in range(100):

    new_value = value.copy()

    for i in range(4):
        for j in range(4):

            if (i,j) == goal:
                continue

            actions = []

            if i > 0:
                actions.append(value[i-1,j])

            if i < 3:
                actions.append(value[i+1,j])

            if j > 0:
                actions.append(value[i,j-1])

            if j < 3:
                actions.append(value[i,j+1])

            new_value[i,j] = -1 + gamma * max(actions)

    value = new_value

print(value)
```
