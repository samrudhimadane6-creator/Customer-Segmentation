```python
import pandas as pd
import matplotlib.pyplot as plt
import gradio as gr
from datetime import datetime
```

```python
expenses = pd.DataFrame(
    columns=["Date", "Category", "Amount", "Description"]
)

monthly_budget = 0
```

```python
def set_budget(budget):
    global monthly_budget

    if budget is None or budget <= 0:
        return "⚠️ Please enter a valid budget."

    monthly_budget = budget

    return f"✅ Monthly Budget Set Successfully: ₹{monthly_budget:,.2f}"


def add_expense(category, amount, date, description):
    global expenses

    if amount is None or amount <= 0:
        return (
            expenses,
            "⚠️ Please enter a valid amount.",
            get_summary(),
            create_chart()
        )

    new_expense = pd.DataFrame({
        "Date": [date],
        "Category": [category],
        "Amount": [amount],
        "Description": [description]
    })

    expenses = pd.concat(
        [expenses, new_expense],
        ignore_index=True
    )

    return (
        expenses,
        f"✅ Expense Added Successfully: ₹{amount}",
        get_summary(),
        create_chart()
    )


def get_summary():

    if expenses.empty:
        return """
        ### 📊 Expense Summary

        💰 **Monthly Budget:** ₹0

        💸 **Total Spent:** ₹0

        🏦 **Remaining Budget:** ₹0

        📌 **Highest Expense Category:** No Data
        """

    total_spent = expenses["Amount"].sum()

    remaining = monthly_budget - total_spent

    category_total = expenses.groupby(
        "Category"
    )["Amount"].sum()

    highest_category = category_total.idxmax()

    summary = f"""
    # 📊 Expense Summary

    💰 **Monthly Budget:** ₹{monthly_budget:,.2f}

    💸 **Total Spent:** ₹{total_spent:,.2f}

    🏦 **Remaining Budget:** ₹{remaining:,.2f}

    📌 **Highest Expense Category:** {highest_category}
    """

    if remaining < 0:
        summary += "\n\n⚠️ **Warning: You have exceeded your budget!**"

    return summary


def create_chart():

    if expenses.empty:
        fig, ax = plt.subplots()
        ax.text(
            0.5,
            0.5,
            "No Expense Data Available",
            ha="center",
            va="center",
            fontsize=16
        )
        ax.axis("off")

        return fig

    category_total = expenses.groupby(
        "Category"
    )["Amount"].sum()

    fig, ax = plt.subplots(figsize=(7, 5))

    ax.pie(
        category_total,
        labels=category_total.index,
        autopct="%1.1f%%",
        startangle=90
    )

    ax.set_title("Category Wise Expense Distribution")

    return fig


def clear_all_expenses():
    global expenses

    expenses = pd.DataFrame(
        columns=[
            "Date",
            "Category",
            "Amount",
            "Description"
        ]
    )

    return (
        expenses,
        "🗑️ All expenses cleared.",
        get_summary(),
        create_chart()
    )
```

```python
!pip install -q gradio pandas matplotlib

import pandas as pd
import matplotlib.pyplot as plt
import gradio as gr
from datetime import datetime


# =========================================================
# DATA
# =========================================================

expenses = pd.DataFrame(
    columns=["Date", "Category", "Amount", "Description"]
)

monthly_budget = 0


# =========================================================
# SET BUDGET
# =========================================================

def set_budget(budget):

    global monthly_budget

    if budget is None or budget <= 0:
        return "⚠️ Please enter a valid budget."

    monthly_budget = float(budget)

    return f"✅ Monthly Budget Set Successfully: ₹{monthly_budget:,.2f}"


# =========================================================
# SUMMARY
# =========================================================

def get_summary():

    global expenses, monthly_budget

    if expenses.empty:
        return f"""
## 📊 Expense Summary

💰 **Monthly Budget:** ₹{monthly_budget:,.2f}

💸 **Total Spent:** ₹0.00

🏦 **Remaining Budget:** ₹{monthly_budget:,.2f}

📌 **Highest Expense Category:** No Data
"""

    total_spent = pd.to_numeric(
        expenses["Amount"],
        errors="coerce"
    ).fillna(0).sum()

    remaining = monthly_budget - total_spent

    category_total = expenses.groupby(
        "Category"
    )["Amount"].sum()

    highest_category = category_total.idxmax()

    summary = f"""
## 📊 Expense Summary

💰 **Monthly Budget:** ₹{monthly_budget:,.2f}

💸 **Total Spent:** ₹{total_spent:,.2f}

🏦 **Remaining Budget:** ₹{remaining:,.2f}

📌 **Highest Expense Category:** {highest_category}
"""

    # Budget exceeded
    if monthly_budget > 0 and total_spent > monthly_budget:

        exceeded = total_spent - monthly_budget

        summary += f"""

## 🚨 ⚠️ BUDGET ALERT!

❌ **You exceeded your budget by ₹{exceeded:,.2f}**

💰 Budget: ₹{monthly_budget:,.2f}

💸 Total Spent: ₹{total_spent:,.2f}
"""

    # Exactly budget
    elif monthly_budget > 0 and total_spent == monthly_budget:

        summary += """

## ⚠️ BUDGET ALERT!

You have used **100% of your monthly budget**.
"""

    # 80% warning
    elif monthly_budget > 0 and total_spent >= monthly_budget * 0.80:

        summary += f"""

### ⚠️ Warning

You have used more than **80% of your budget**.

Only **₹{remaining:,.2f}** is remaining.
"""

    return summary


# =========================================================
# CHART
# =========================================================

def create_chart():

    global expenses

    if expenses.empty:

        fig, ax = plt.subplots(figsize=(7, 5))

        ax.text(
            0.5,
            0.5,
            "No Expense Data Available",
            ha="center",
            va="center",
            fontsize=16
        )

        ax.axis("off")

        return fig

    data = expenses.copy()

    data["Amount"] = pd.to_numeric(
        data["Amount"],
        errors="coerce"
    ).fillna(0)

    category_total = data.groupby(
        "Category"
    )["Amount"].sum()

    fig, ax = plt.subplots(figsize=(7, 5))

    ax.pie(
        category_total.values,
        labels=category_total.index,
        autopct="%1.1f%%",
        startangle=90
    )

    ax.set_title(
        "Category Wise Expense Distribution"
    )

    return fig


# =========================================================
# ADD EXPENSE
# =========================================================

def add_expense(category, amount, date, description):

    global expenses

    # Validate category
    if not category:

        return (
            expenses.copy(),
            "⚠️ Please select an expense category.",
            get_summary(),
            create_chart()
        )

    # Validate amount
    if amount is None or amount <= 0:

        return (
            expenses.copy(),
            "⚠️ Please enter a valid amount.",
            get_summary(),
            create_chart()
        )

    # Create new expense
    new_expense = pd.DataFrame({
        "Date": [date],
        "Category": [category],
        "Amount": [float(amount)],
        "Description": [description if description else ""]
    })

    # =====================================================
    # ADD NEW ROW TO OLD EXPENSES
    # =====================================================

    expenses = pd.concat(
        [
            expenses,
            new_expense
        ],
        ignore_index=True
    )

    # Make sure columns remain correct
    expenses = expenses[
        [
            "Date",
            "Category",
            "Amount",
            "Description"
        ]
    ]

    # Calculate total
    total_spent = expenses["Amount"].sum()

    # Status
    if monthly_budget > 0 and total_spent > monthly_budget:

        exceeded = total_spent - monthly_budget

        status = f"""
🚨 **BUDGET EXCEEDED!**

✅ Expense Added: ₹{amount:,.2f}

💸 Total Spent: ₹{total_spent:,.2f}

💰 Budget: ₹{monthly_budget:,.2f}

❌ Exceeded By: ₹{exceeded:,.2f}
"""

    else:

        status = (
            f"✅ **{category}** expense added successfully: "
            f"₹{amount:,.2f}"
        )

    # Return updated table
    return (
        expenses.copy(),
        status,
        get_summary(),
        create_chart()
    )


# =========================================================
# CLEAR ALL EXPENSES
# =========================================================

def clear_all_expenses():

    global expenses

    expenses = pd.DataFrame(
        columns=[
            "Date",
            "Category",
            "Amount",
            "Description"
        ]
    )

    return (
        expenses.copy(),
        "🗑️ All expenses cleared.",
        get_summary(),
        create_chart()
    )


# =========================================================
# GRADIO APP
# =========================================================

with gr.Blocks(
    title="AI Personal Expense Tracker"
) as app:

    gr.Markdown(
        """
# 💰 AI Personal Expense Tracker

### Manage your daily expenses and monthly budget easily!
"""
    )

    # =====================================================
    # INPUT SECTION
    # =====================================================

    with gr.Row():

        # ---------------- BUDGET ----------------

        with gr.Column():

            gr.Markdown("## 💰 Set Monthly Budget")

            budget_input = gr.Number(
                label="Enter Monthly Budget (₹)",
                placeholder="Example: 10000"
            )

            budget_button = gr.Button(
                "Set Budget",
                variant="primary"
            )

            budget_status = gr.Textbox(
                label="Budget Status"
            )

        # ---------------- EXPENSE ----------------

        with gr.Column():

            gr.Markdown("## ➕ Add New Expense")

            category_input = gr.Dropdown(
                choices=[
                    "Food",
                    "Travel",
                    "Shopping",
                    "Education",
                    "Bills",
                    "Entertainment",
                    "Health",
                    "Other"
                ],
                label="Expense Category",
                value="Food"
            )

            amount_input = gr.Number(
                label="Amount (₹)",
                placeholder="Example: 500"
            )

            date_input = gr.Textbox(
                label="Date",
                value=datetime.now().strftime("%Y-%m-%d")
            )

            description_input = gr.Textbox(
                label="Description",
                placeholder="Example: Lunch, Bus ticket..."
            )

            add_button = gr.Button(
                "➕ Add Expense",
                variant="primary"
            )

    # =====================================================
    # SUMMARY
    # =====================================================

    gr.Markdown("## 📊 Expense Summary")

    summary_output = gr.Markdown(
        get_summary()
    )

    # =====================================================
    # EXPENSE HISTORY
    # =====================================================

    gr.Markdown("## 📋 Expense History")

    expense_table = gr.Dataframe(
        value=expenses,
        headers=[
            "Date",
            "Category",
            "Amount",
            "Description"
        ],
        datatype=[
            "str",
            "str",
            "number",
            "str"
        ],
        label="All Added Expenses",
        interactive=False
    )

    # =====================================================
    # CHART
    # =====================================================

    gr.Markdown("## 📈 Expense Analysis")

    expense_chart = gr.Plot(
        value=create_chart()
    )

    # =====================================================
    # STATUS
    # =====================================================

    status_output = gr.Markdown()

    # =====================================================
    # CLEAR BUTTON
    # =====================================================

    clear_button = gr.Button(
        "🗑️ Clear All Expenses"
    )

    # =====================================================
    # SET BUDGET EVENT
    # =====================================================

    budget_button.click(
        set_budget,
        inputs=[
            budget_input
        ],
        outputs=[
            budget_status
        ]
    ).then(
        get_summary,
        inputs=[],
        outputs=[
            summary_output
        ]
    )

    # =====================================================
    # ADD EXPENSE EVENT
    # =====================================================

    add_button.click(
        add_expense,
        inputs=[
            category_input,
            amount_input,
            date_input,
            description_input
        ],
        outputs=[
            expense_table,
            status_output,
            summary_output,
            expense_chart
        ]
    )

    # =====================================================
    # CLEAR EVENT
    # =====================================================

    clear_button.click(
        clear_all_expenses,
        inputs=[],
        outputs=[
            expense_table,
            status_output,
            summary_output,
            expense_chart
        ]
    )


# =========================================================
# LAUNCH
# =========================================================

app.launch(
    share=True,
    debug=False
)
```
