```python
import pandas as pd
import matplotlib.pyplot as plt
import gradio as gr
from datetime import datetime
```

```python
expenses = pd.DataFrame(
    columns=["Date", "Category", "Amount", "Description"]
)

monthly_budget = 0
```

```python
def set_budget(budget):
    global monthly_budget

    if budget is None or budget <= 0:
        return "⚠️ Please enter a valid budget."

    monthly_budget = budget

    return f"✅ Monthly Budget Set Successfully: ₹{monthly_budget:,.2f}"


def add_expense(category, amount, date, description):
    global expenses

    if amount is None or amount <= 0:
        return (
            expenses,
            "⚠️ Please enter a valid amount.",
            get_summary(),
            create_chart()
        )

    new_expense = pd.DataFrame({
        "Date": [date],
        "Category": [category],
        "Amount": [amount],
        "Description": [description]
    })

    expenses = pd.concat(
        [expenses, new_expense],
        ignore_index=True
    )

    return (
        expenses,
        f"✅ Expense Added Successfully: ₹{amount}",
        get_summary(),
        create_chart()
    )


def get_summary():

    if expenses.empty:
        return """
        ### 📊 Expense Summary

        💰 **Monthly Budget:** ₹0

        💸 **Total Spent:** ₹0

        🏦 **Remaining Budget:** ₹0

        📌 **Highest Expense Category:** No Data
        """

    total_spent = expenses["Amount"].sum()

    remaining = monthly_budget - total_spent

    category_total = expenses.groupby(
        "Category"
    )["Amount"].sum()

    highest_category = category_total.idxmax()

    summary = f"""
    # 📊 Expense Summary

    💰 **Monthly Budget:** ₹{monthly_budget:,.2f}

    💸 **Total Spent:** ₹{total_spent:,.2f}

    🏦 **Remaining Budget:** ₹{remaining:,.2f}

    📌 **Highest Expense Category:** {highest_category}
    """

    if remaining < 0:
        summary += "\n\n⚠️ **Warning: You have exceeded your budget!**"

    return summary


def create_chart():

    if expenses.empty:
        fig, ax = plt.subplots()
        ax.text(
            0.5,
            0.5,
            "No Expense Data Available",
            ha="center",
            va="center",
            fontsize=16
        )
        ax.axis("off")

        return fig

    category_total = expenses.groupby(
        "Category"
    )["Amount"].sum()

    fig, ax = plt.subplots(figsize=(7, 5))

    ax.pie(
        category_total,
        labels=category_total.index,
        autopct="%1.1f%%",
        startangle=90
    )

    ax.set_title("Category Wise Expense Distribution")

    return fig


def clear_all_expenses():
    global expenses

    expenses = pd.DataFrame(
        columns=[
            "Date",
            "Category",
            "Amount",
            "Description"
        ]
    )

    return (
        expenses,
        "🗑️ All expenses cleared.",
        get_summary(),
        create_chart()
    )
```

```python
with gr.Blocks(
    title="AI Personal Expense Tracker"
) as app:

    gr.Markdown(
        """
        # 💰 AI Personal Expense Tracker

        ### Manage your daily expenses and monthly budget easily!
        """
    )

    with gr.Row():

        with gr.Column():

            gr.Markdown("## 💰 Set Monthly Budget")

            budget_input = gr.Number(
                label="Enter Monthly Budget (₹)",
                placeholder="Example: 10000"
            )

            budget_button = gr.Button(
                "Set Budget"
            )

            budget_status = gr.Textbox(
                label="Budget Status"
            )

        with gr.Column():

            gr.Markdown("## ➕ Add New Expense")

            category_input = gr.Dropdown(
                choices=[
                    "Food",
                    "Travel",
                    "Shopping",
                    "Education",
                    "Bills",
                    "Entertainment",
                    "Health",
                    "Other"
                ],
                label="Expense Category",
                value="Food"
            )

            amount_input = gr.Number(
                label="Amount (₹)",
                placeholder="Example: 500"
            )

            date_input = gr.Textbox(
                label="Date",
                value=datetime.now().strftime(
                    "%Y-%m-%d"
                )
            )

            description_input = gr.Textbox(
                label="Description",
                placeholder="Example: Lunch, Bus ticket..."
            )

            add_button = gr.Button(
                "➕ Add Expense"
            )

    gr.Markdown("## 📊 Expense Summary")

    summary_output = gr.Markdown(
        get_summary()
    )

    gr.Markdown("## 📋 Expense History")

    expense_table = gr.Dataframe(
        value=expenses,
        label="Your Expenses",
        interactive=False
    )

    gr.Markdown("## 📈 Expense Analysis")

    expense_chart = gr.Plot(
        value=create_chart()
    )

    with gr.Row():

        clear_button = gr.Button(
            "🗑️ Clear All Expenses"
        )

        status_output = gr.Textbox(
            label="Application Status"
        )

    # Set Budget Button
    budget_button.click(
        set_budget,
        inputs=[
            budget_input
        ],
        outputs=[
            budget_status
        ]
    )

    # Add Expense Button
    add_button.click(
        add_expense,
        inputs=[
            category_input,
            amount_input,
            date_input,
            description_input
        ],
        outputs=[
            expense_table,
            status_output,
            summary_output,
            expense_chart
        ]
    )

    # Clear Button
    clear_button.click(
        clear_all_expenses,
        outputs=[
            expense_table,
            status_output,
            summary_output,
            expense_chart
        ]
    )
```

```python
app.launch(share=True)
```
