```python
import numpy as np

V = np.zeros((5,5))

alpha = 0.1
gamma = 0.9

for episode in range(1000):

    state = [0,0]

    done = False

    while not done:

        action = np.random.randint(4)

        next_state = state.copy()

        if action == 0:
            next_state[0] = max(0, next_state[0]-1)

        elif action == 1:
            next_state[0] = min(4, next_state[0]+1)

        elif action == 2:
            next_state[1] = max(0, next_state[1]-1)

        else:
            next_state[1] = min(4, next_state[1]+1)

        reward = -1

        if next_state == [4,4]:
            reward = 10
            done = True

        V[state[0], state[1]] += alpha * (
            reward + gamma * V[next_state[0], next_state[1]]
            - V[state[0], state[1]]
        )

        state = next_state

print(V)
```
