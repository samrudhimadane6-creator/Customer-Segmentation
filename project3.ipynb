```python
import gymnasium as gym
```

```python
env = gym.make("Taxi-v4", render_mode="ansi")
```

```python
observation, info = env.reset()
```

```python
print(env.render())
```

```python
action = env.action_space.sample()

print("Game Started")
```

```python
print("Action:", action)
```

```python
observation, reward, terminated, truncated, info = env.step(action)
```

```python
print("Reward:", reward)
```

```python
print(env.render())
```

```python
observation, info = env.reset()

for i in range(10):

    action = env.action_space.sample()

    observation, reward, terminated, truncated, info = env.step(action)

    print(env.render())

    print("Step:", i + 1)

    print("Reward:", reward)

    if terminated or truncated:
        break
```
