```python
import numpy as np
import matplotlib.pyplot as plt

epsilon = 0.2

exploration = []
exploitation = []

for i in range(1000):

    if np.random.rand() < epsilon:
        exploration.append(1)
        exploitation.append(0)
    else:
        exploration.append(0)
        exploitation.append(1)

plt.plot(np.cumsum(exploration), label="Exploration")
plt.plot(np.cumsum(exploitation), label="Exploitation")

plt.legend()
plt.title("Epsilon Greedy Policy")
plt.show()
```
