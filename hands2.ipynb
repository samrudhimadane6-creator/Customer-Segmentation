```python
import numpy as np
import matplotlib.pyplot as plt

# Rewards probability
bandits = [0.2, 0.5, 0.75, 0.3]

Q = np.zeros(4)
N = np.zeros(4)

epsilon = 0.1
rewards = []

for i in range(1000):

    # Exploration
    if np.random.rand() < epsilon:
        action = np.random.randint(4)

    # Exploitation
    else:
        action = np.argmax(Q)

    reward = 1 if np.random.rand() < bandits[action] else 0

    N[action] += 1

    Q[action] = Q[action] + (1 / N[action]) * (reward - Q[action])

    rewards.append(reward)

print("Learned Values:", Q)

plt.plot(np.cumsum(rewards))
plt.xlabel("Rounds")
plt.ylabel("Total Rewards")
plt.title("Multi Armed Bandit Learning")
plt.show()
```
