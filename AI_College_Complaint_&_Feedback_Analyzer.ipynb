```python
import pandas as pd
import joblib

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

# Load dataset
df = pd.read_csv("complaints.csv")

print("Dataset loaded successfully!")
print("Total complaints:", len(df))


# ==============================
# 1. CATEGORY MODEL
# ==============================

category_model = Pipeline([
    ("tfidf", TfidfVectorizer(
        lowercase=True,
        stop_words="english",
        ngram_range=(1, 2)
    )),
    ("classifier", LogisticRegression(
        max_iter=1000
    ))
])

category_model.fit(
    df["complaint"],
    df["category"]
)

# Save category model
joblib.dump(
    category_model,
    "category_model.pkl"
)


# ==============================
# 2. SENTIMENT MODEL
# ==============================

sentiment_model = Pipeline([
    ("tfidf", TfidfVectorizer(
        lowercase=True,
        stop_words="english",
        ngram_range=(1, 2)
    )),
    ("classifier", LogisticRegression(
        max_iter=1000
    ))
])

sentiment_model.fit(
    df["complaint"],
    df["sentiment"]
)

# Save sentiment model
joblib.dump(
    sentiment_model,
    "sentiment_model.pkl"
)


print("\n✅ Category model trained successfully!")
print("✅ Sentiment model trained successfully!")
print("✅ Models saved successfully!")
```

```python
import gradio as gr
import pandas as pd
import joblib
import os
from datetime import datetime

# =========================================================
# LOAD MODELS
# =========================================================

category_model = joblib.load("category_model.pkl")
sentiment_model = joblib.load("sentiment_model.pkl")

FILE_NAME = "college_complaints.csv"


# =========================================================
# PRIORITY FUNCTION
# =========================================================

def get_priority(text, sentiment):

    urgent_words = [
        "emergency",
        "danger",
        "unsafe",
        "fire",
        "accident",
        "electric shock",
        "security",
        "harassment",
        "not working",
        "broken"
    ]

    text = text.lower()

    if any(word in text for word in urgent_words):
        return "High"

    elif sentiment == "Negative":
        return "Medium"

    return "Low"


# =========================================================
# CREATE EMPTY DATABASE IF NOT EXISTS
# =========================================================

columns = [
    "Complaint_ID",
    "Student_Name",
    "Student_ID",
    "Complaint",
    "Category",
    "Sentiment",
    "Priority",
    "Status",
    "Date_Time"
]

if not os.path.exists(FILE_NAME):

    pd.DataFrame(columns=columns).to_csv(
        FILE_NAME,
        index=False
    )


# =========================================================
# SUBMIT COMPLAINT
# =========================================================

def submit_complaint(student_name, student_id, complaint):

    if not student_name.strip():
        return "⚠️ Please enter Student Name."

    if not student_id.strip():
        return "⚠️ Please enter Student ID."

    if not complaint.strip():
        return "⚠️ Please enter your complaint."

    df = pd.read_csv(FILE_NAME)

    # Generate Complaint ID
    complaint_number = len(df) + 1
    complaint_id = f"CMP{complaint_number:04d}"

    # AI Prediction
    category = category_model.predict([complaint])[0]
    sentiment = sentiment_model.predict([complaint])[0]

    priority = get_priority(
        complaint,
        sentiment
    )

    date_time = datetime.now().strftime(
        "%Y-%m-%d %H:%M:%S"
    )

    # New record
    new_record = pd.DataFrame([{
        "Complaint_ID": complaint_id,
        "Student_Name": student_name,
        "Student_ID": student_id,
        "Complaint": complaint,
        "Category": category,
        "Sentiment": sentiment,
        "Priority": priority,
        "Status": "Pending",
        "Date_Time": date_time
    }])

    df = pd.concat(
        [df, new_record],
        ignore_index=True
    )

    df.to_csv(
        FILE_NAME,
        index=False
    )

    return f"""
# ✅ Complaint Submitted Successfully!

### 🎫 Complaint ID
## {complaint_id}

### 🤖 AI Analysis

**Category:** {category}

**Sentiment:** {sentiment}

**Priority:** {priority} 🚨

**Status:** Pending

### 🕐 Submitted At

{date_time}

---

📌 **Save your Complaint ID to track your complaint.**
"""


# =========================================================
# SHOW ALL COMPLAINTS
# =========================================================

def show_all_complaints():

    df = pd.read_csv(FILE_NAME)

    if df.empty:
        return pd.DataFrame(
            columns=columns
        )

    return df


# =========================================================
# TRACK COMPLAINT
# =========================================================

def track_complaint(complaint_id):

    if not complaint_id.strip():
        return "⚠️ Please enter Complaint ID."

    df = pd.read_csv(FILE_NAME)

    complaint_id = complaint_id.strip().upper()

    result = df[
        df["Complaint_ID"].str.upper() == complaint_id
    ]

    if result.empty:

        return """
# ❌ Complaint Not Found

Please check your Complaint ID.

Example:

`CMP0001`
"""

    row = result.iloc[0]

    return f"""
# 🎫 Complaint Details

### Complaint ID
**{row['Complaint_ID']}**

### Student
**{row['Student_Name']}**

### Complaint
**{row['Complaint']}**

### Category
**{row['Category']}**

### Sentiment
**{row['Sentiment']}**

### Priority
**{row['Priority']}**

### Current Status
## {row['Status']}

### Submitted At
{row['Date_Time']}
"""


# =========================================================
# UPDATE COMPLAINT STATUS
# =========================================================

def update_status(complaint_id, new_status):

    if not complaint_id.strip():
        return "⚠️ Please enter Complaint ID."

    df = pd.read_csv(FILE_NAME)

    complaint_id = complaint_id.strip().upper()

    index = df[
        df["Complaint_ID"].str.upper() == complaint_id
    ].index

    if len(index) == 0:

        return "❌ Complaint ID not found."

    df.loc[index, "Status"] = new_status

    df.to_csv(
        FILE_NAME,
        index=False
    )

    return f"""
# ✅ Status Updated

**Complaint ID:** {complaint_id}

**New Status:** {new_status}
"""


# =========================================================
# DASHBOARD SUMMARY
# =========================================================

def dashboard_summary():

    df = pd.read_csv(FILE_NAME)

    total = len(df)

    pending = len(
        df[df["Status"] == "Pending"]
    )

    progress = len(
        df[df["Status"] == "In Progress"]
    )

    resolved = len(
        df[df["Status"] == "Resolved"]
    )

    high = len(
        df[df["Priority"] == "High"]
    )

    return f"""
## 📊 Complaint Dashboard

| Metric | Count |
|---|---:|
| 📋 Total Complaints | **{total}** |
| 🟡 Pending | **{pending}** |
| 🔵 In Progress | **{progress}** |
| 🟢 Resolved | **{resolved}** |
| 🚨 High Priority | **{high}** |
"""


# =========================================================
# GRADIO FRONTEND
# =========================================================

with gr.Blocks(
    title="AI College Complaint Analyzer"
) as demo:

    gr.Markdown("""
# 🎓 AI College Complaint & Feedback Analyzer

### Smart College Complaint Management System

🤖 **AI Category Detection**
😊 **Sentiment Analysis**
🚨 **Priority Detection**
🎫 **Complaint Tracking**
📊 **Admin Dashboard**
""")


    # =====================================================
    # TAB 1 - STUDENT COMPLAINT
    # =====================================================

    with gr.Tab("📝 Submit Complaint"):

        gr.Markdown(
            "### Enter your complaint"
        )

        student_name = gr.Textbox(
            label="👤 Student Name",
            placeholder="Enter your full name"
        )

        student_id = gr.Textbox(
            label="🎓 Student ID",
            placeholder="Example: AIML101"
        )

        complaint = gr.Textbox(
            label="📝 Complaint / Feedback",
            placeholder=(
                "Example: The classroom fans are not working..."
            ),
            lines=6
        )

        submit_button = gr.Button(
            "🚀 Analyze & Submit",
            variant="primary"
        )

        submit_result = gr.Markdown()

        submit_button.click(
            fn=submit_complaint,
            inputs=[
                student_name,
                student_id,
                complaint
            ],
            outputs=submit_result
        )


    # =====================================================
    # TAB 2 - TRACK COMPLAINT
    # =====================================================

    with gr.Tab("🔎 Track Complaint"):

        gr.Markdown("""
### 🔎 Track Your Complaint

Enter the Complaint ID received after submission.
""")

        track_id = gr.Textbox(
            label="🎫 Complaint ID",
            placeholder="Example: CMP0001"
        )

        track_button = gr.Button(
            "🔍 Track Complaint"
        )

        track_result = gr.Markdown()

        track_button.click(
            fn=track_complaint,
            inputs=track_id,
            outputs=track_result
        )


    # =====================================================
    # TAB 3 - ADMIN DASHBOARD
    # =====================================================

    with gr.Tab("📊 Admin Dashboard"):

        gr.Markdown("""
# 👨‍💼 Admin Dashboard

College administration can view and manage all complaints.
""")

        summary_button = gr.Button(
            "📊 Show Dashboard Summary"
        )

        summary_result = gr.Markdown()

        summary_button.click(
            fn=dashboard_summary,
            inputs=None,
            outputs=summary_result
        )


        gr.Markdown(
            "## 📋 All Complaints"
        )

        show_button = gr.Button(
            "👁️ Show All Complaints"
        )

        complaint_table = gr.Dataframe(
            interactive=False
        )

        show_button.click(
            fn=show_all_complaints,
            inputs=None,
            outputs=complaint_table
        )


        # =================================================
        # UPDATE STATUS
        # =================================================

        gr.Markdown(
            "## 🔄 Update Complaint Status"
        )

        admin_complaint_id = gr.Textbox(
            label="🎫 Complaint ID",
            placeholder="Example: CMP0001"
        )

        status_dropdown = gr.Dropdown(
            choices=[
                "Pending",
                "In Progress",
                "Resolved"
            ],
            label="Select New Status",
            value="Pending"
        )

        update_button = gr.Button(
            "🔄 Update Status",
            variant="primary"
        )

        update_result = gr.Markdown()

        update_button.click(
            fn=update_status,
            inputs=[
                admin_complaint_id,
                status_dropdown
            ],
            outputs=update_result
        )


    # =====================================================
    # FOOTER
    # =====================================================

    gr.Markdown("""
---

### 🎓 AI College Complaint Management System

**Python | NLP | TF-IDF | Machine Learning | Gradio**

💡 Designed for real-world college complaint management.
""")


# =========================================================
# LAUNCH
# =========================================================

demo.launch(share=True)
```
